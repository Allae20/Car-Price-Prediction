```python
from selenium.webdriver.common.by import By

def extraire_annonce(driver):
    data = {}

    rows = driver.find_elements(By.TAG_NAME, "tr")

    for row in rows:
        texte = row.text.strip()

        if ":" in texte:
            parties = texte.split()

            # extraction simple par mots clés
            if "Marque:" in texte:
                data["Marque"] = texte.split("Marque:")[1].split()[0]

            if "Modèle:" in texte:
                data["Modele"] = texte.split("Modèle:")[1].split()[0]

            if "Année:" in texte:
                data["Annee"] = texte.split("Année:")[1].split()[0]

            if "Kilométrage:" in texte:
                data["Kilometrage"] = texte.split("Kilométrage:")[1].split()[0]

            if "Carburant:" in texte:
                data["Carburant"] = texte.split("Carburant:")[1].split()[0]

            if "Transmission:" in texte:
                data["Transmission"] = texte.split("Transmission:")[1].split()[0]

    return data
```

```python
from selenium.webdriver.common.by import By
import time
import re

def extraire_annonce(url):
    driver.get(url)
    time.sleep(2)

    data = {}

    # Titre (marque + modèle)
    try:
        titre = driver.find_element(By.TAG_NAME, "h1").text
        data["Titre"] = titre
    except:
        data["Titre"] = "N/A"


    # Texte complet
    texte = driver.find_element(By.TAG_NAME, "body").text


    # Année
    match = re.search(r"Année:\s*(\d{4})", texte)
    data["Annee"] = match.group(1) if match else "N/A"


    # Kilométrage
    match = re.search(r"Kilométrage:\s*([\d,\s]+)", texte)
    data["Kilometrage"] = match.group(1).strip() if match else "N/A"


    # Carburant
    match = re.search(r"Carburant:\s*(\w+)", texte)
    data["Carburant"] = match.group(1) if match else "N/A"


    # Transmission
    match = re.search(r"Transmission:\s*(\w+)", texte)
    data["Transmission"] = match.group(1) if match else "N/A"


    return data
```

```python
from selenium.webdriver.common.by import By
import pandas as pd
import re
import time

donnees = []

for i, url in enumerate(urls):

    print(f"{i+1}/{len(urls)}")

    try:
        driver.get(url)
        time.sleep(2)

        texte = driver.find_element(By.TAG_NAME, "body").text

        # Prix
        prix = ""
        m = re.search(r"Comparer\s*×.*?([\d\s,]+)\s*MAD", texte, re.S)
        if m:
            prix = m.group(1).strip().replace(" ", "")

        if prix == "":
            m = re.search(r"Prix sur demande", texte)
            if m:
                prix = "Prix sur demande"

        # Année
        annee = ""
        m = re.search(r"Année:\s*(\d{4})", texte)
        if m:
            annee = m.group(1)

        # Kilométrage
        km = ""
        m = re.search(r"Kilométrage:\s*([\d\s,]+)", texte)
        if m:
            km = m.group(1).strip().replace(" ", "")

        # Carburant
        carburant = ""
        m = re.search(r"Carburant:\s*([^\n]+)", texte)
        if m:
            carburant = m.group(1).strip()

        # Transmission
        transmission = ""
        m = re.search(r"Transmission:\s*([^\n]+)", texte)
        if m:
            transmission = m.group(1).strip()

        # Marque
        marque = ""
        m = re.search(r"Marque:\s*([^\n]+)", texte)
        if m:
            marque = m.group(1).strip()

        # Modèle
        modele = ""
        m = re.search(r"Modèle:\s*([^\n]+)", texte)
        if m:
            modele = m.group(1).strip()

        donnees.append({
            "Prix": prix,
            "Marque": marque,
            "Modele": modele,
            "Annee": annee,
            "Kilometrage": km,
            "Carburant": carburant,
            "Transmission": transmission,
            "URL": url
        })

    except Exception as e:
        print("Erreur :", e)
```

```python
def extraire_details(url):

    driver.get(url)
    time.sleep(2)

    details = {
        "annee": None,
        "kilometrage": None,
        "carburant": None,
        "boite": None
    }

    # récupérer les td
    tds = driver.find_elements(By.TAG_NAME, "td")

    valeurs = [td.text.strip() for td in tds if td.text.strip()]

    for v in valeurs:
        if v.isdigit() and len(v) == 4:
            details["annee"] = v

        elif "km" in v:
            details["kilometrage"] = v

        elif v in ["Diesel", "Essence", "Hybride", "Electrique"]:
            details["carburant"] = v

    # description
    texte = driver.find_element(By.TAG_NAME,"body").text

    if "Boîte de vitesses :" in texte:
        boite = texte.split("Boîte de vitesses :")[1].split("-")[0]
        details["boite"] = boite.strip()

    return details
```

```python
def extraire_annonce_rapide(url):
    driver.get(url)
    time.sleep(1)

    annonce = {
        "URL": url,
        "Titre": None,
        "Annee": None,
        "Kilometrage": None,
        "Carburant": None,
        "Transmission": None
    }

    texte = driver.find_element(By.TAG_NAME, "body").text

    lignes = [x.strip() for x in texte.split("\n") if x.strip()]

    # Titre
    if "Acheter une voiture d'occasion" in lignes:
        try:
            idx = lignes.index("Acheter une voiture d'occasion")
            annonce["Titre"] = lignes[idx+1]
        except:
            pass

    # Année
    for ligne in lignes:
        if ligne.isdigit() and len(ligne) == 4:
            annonce["Annee"] = ligne
            break

    # Kilométrage
    for ligne in lignes:
        if "km" in ligne:
            annonce["Kilometrage"] = ligne.replace("km","").strip()
            break

    # Carburant
    for c in ["Diesel", "Essence", "Hybride", "Electrique"]:
        if c in texte:
            annonce["Carburant"] = c
            break

    # Transmission
    if "Automatique" in texte:
        annonce["Transmission"] = "Automatique"
    elif "Manuelle" in texte:
        annonce["Transmission"] = "Manuelle"

    return annonce
```

```python
import re
import pandas as pd

def extraire_infos(texte):

    voiture = {}

    # Marque + Modèle
    m = re.search(r"Marque:\s*(.*?)\s+Modèle:\s*(.*?)\n", texte)
    if m:
        voiture["Marque"] = m.group(1).strip()
        voiture["Modele"] = m.group(2).strip()
    else:
        voiture["Marque"] = None
        voiture["Modele"] = None

    # Année
    m = re.search(r"Année:\s*(\d{4})", texte)
    voiture["Annee"] = m.group(1) if m else None

    # Kilométrage
    m = re.search(r"Kilométrage:\s*([\d,\s]+)\s*km", texte)
    voiture["Kilometrage"] = m.group(1).strip() if m else None

    # Carburant
    m = re.search(r"Carburant:\s*(Diesel|Essence|Hybride|Electrique)", texte)
    voiture["Carburant"] = m.group(1) if m else None

    # Transmission
    m = re.search(r"Transmission:\s*(Automatique|Manuelle)", texte)
    voiture["Transmission"] = m.group(1) if m else None

    # Prix
    m = re.search(r"(\d{1,3}(?:,\d{3})*)\s*MAD", texte)
    voiture["Prix"] = m.group(1).replace(",","") if m else None

    return voiture


# Application sur les 125 annonces
resultats = []

for texte in df_125["Texte"]:
    resultats.append(extraire_infos(texte))


df_final = pd.DataFrame(resultats)

df_final.head()
```

```python
import re
import pandas as pd

def extraire_infos(row):
    texte = row["Texte"]

    resultat = {}

    # URL
    resultat["URL"] = row["URL"]

    # Marque + modèle
    m = re.search(r"Marque:\s*(.*?)\s+Modèle:\s*(.*)", texte)
    if m:
        resultat["Marque"] = m.group(1)
        resultat["Modele"] = m.group(2).split("\n")[0]
    else:
        resultat["Marque"] = None
        resultat["Modele"] = None

    # Année
    annee = re.search(r"Année:\s*(\d{4})", texte)
    resultat["Annee"] = annee.group(1) if annee else None

    # Kilométrage
    km = re.search(r"Kilométrage:\s*([\d\s,]+)", texte)
    resultat["Kilometrage"] = km.group(1).replace(",", "") if km else None

    # Carburant
    carburant = re.search(r"Carburant:\s*(\w+)", texte)
    resultat["Carburant"] = carburant.group(1) if carburant else None

    # Transmission
    boite = re.search(r"Transmission:\s*(\w+)", texte)
    resultat["Transmission"] = boite.group(1) if boite else None

    return resultat


df_final = df_brut.apply(extraire_infos, axis=1, result_type="expand")

df_final.head()
```

```python
def extraire_prix(texte):
    prix = re.search(r"(\d{1,3}(?:,\d{3})*)\s*MAD", texte)

    if prix:
        return int(prix.group(1).replace(",", ""))
    else:
        return None


df_final["Prix"] = df_brut["Texte"].apply(extraire_prix)

df_final.head()
```
